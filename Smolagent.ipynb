```python
import sys
import subprocess

# Install ddgs into the SAME Python environment used by Jupyter
subprocess.check_call([
    sys.executable, "-m", "pip", "install", "ddgs"
])

# Import
from smolagents import CodeAgent, DuckDuckGoSearchTool

# Define the ML Code Generation Agent
from smolagents import CodeAgent

ml_code_agent = CodeAgent(
    tools=[],
    model=model,
    max_steps=2,
    additional_authorized_imports=[
        "pandas",
        "numpy",
        "sklearn",
        "json"
    ]
)

print("✅ Agent ready!")
```

```python
# function to generate ML code step by step
def generate_ml_code(prompt):
    full_prompt = f"""create a step by step plan to write python code for svm model using sklearn to classify the iris dataset."""

    try:
        response = ml_code_agent.run(full_prompt)
        return response
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
# gradio UI for User Interaction
import gradio as gr
# Create a Gradio interface
iface = gr.Interface(
    fn=generate_ml_code,
    inputs= "text",
    outputs="text", 
    title="ML Code Generation Agent(Created by Ankita)",
    description="Enter a prompt to generate required user information",
)
```

```python
# Lanch Gradio App
iface.launch(share=True, debug=True)
```
